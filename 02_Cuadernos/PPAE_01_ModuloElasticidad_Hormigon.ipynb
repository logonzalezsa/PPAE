## 1. Contexto ingenieril

El módulo de elasticidad del hormigón, $E_c$, es una propiedad mecánica que permite caracterizar su respuesta ante la aplicación de esfuerzos dentro del intervalo de comportamiento considerado para su evaluación.

Su valor está relacionado con diferentes características del hormigón y, entre ellas, con las propiedades del agregado grueso utilizado en su elaboración. Cuando no se dispone del valor de la masa unitaria del hormigón, pueden emplearse expresiones empíricas para estimar $E_c$ en función de la resistencia especificada a la compresión, $f'_c$, considerando el origen del agregado grueso.

En este cuaderno se estudiarán cuatro condiciones:

- Agregado grueso de origen ígneo.
- Agregado grueso de origen metamórfico.
- Agregado grueso de origen sedimentario.
- Origen del agregado grueso sin distinción.

### Objetivo del cuaderno

Implementar un procedimiento computacional para estimar el módulo de elasticidad del hormigón, $E_c$, en función de $f'_c$ y del origen del agregado grueso, almacenar los resultados obtenidos y representar gráficamente su comportamiento para facilitar su análisis e interpretación ingenieril.

## 2. Formulación del problema

Cuando no se dispone del valor de la masa unitaria del hormigón, el módulo de elasticidad, $E_c$, puede estimarse mediante expresiones que dependen de la resistencia especificada a la compresión, $f'_c$, y del origen del agregado grueso.

Las expresiones consideradas son:

**Agregado grueso de origen ígneo:**

$$
E_c = 5500\sqrt{f'_c}
$$

**Agregado grueso de origen metamórfico:**

$$
E_c = 4700\sqrt{f'_c}
$$

**Agregado grueso de origen sedimentario:**

$$
E_c = 3600\sqrt{f'_c}
$$

**Sin distinción del origen del agregado grueso:**

$$
E_c = 3900\sqrt{f'_c}
$$

donde:

- $E_c$: módulo de elasticidad del hormigón [MPa].
- $f'_c$: resistencia especificada a la compresión del hormigón [MPa].

Para representar computacionalmente el origen del agregado grueso se utilizará la variable **OA**, definida de la siguiente manera:

| OA | Origen del agregado grueso |
|---:|:----------------------------|
| 0 | Ígneo |
| 1 | Metamórfico |
| 2 | Sedimentario |
| 3 | Sin distinción |

## 3. Formulación computacional

Las cuatro expresiones utilizadas para estimar el módulo de elasticidad del hormigón presentan la misma estructura matemática y pueden representarse mediante una expresión general:

$$
E_c = K\sqrt{f'_c}
$$

donde el valor de la constante $K$ depende del origen del agregado grueso:

| OA | Origen del agregado grueso | K |
|---:|:----------------------------|---:|
| 0 | Ígneo | 5500 |
| 1 | Metamórfico | 4700 |
| 2 | Sedimentario | 3600 |
| 3 | Sin distinción | 3900 |

Por lo tanto, el procedimiento computacional debe identificar el valor de **OA**, asignar el valor correspondiente de **K** y posteriormente calcular $E_c$ para los diferentes valores de $f'_c$.

Los resultados serán almacenados en una matriz denominada **Ec**, en la cual:

- Las **filas** representan el origen del agregado grueso, identificado mediante **OA**.
- Las **columnas** representan los valores de $f'_c$.
- Cada posición **Ec[OA, J]** almacenará el módulo de elasticidad calculado para una combinación determinada de origen del agregado y resistencia a la compresión.

## 4. Importación de bibliotecas

Para implementar el procedimiento computacional se utilizarán dos bibliotecas de Python:

- **NumPy (`numpy`)**: proporciona herramientas para realizar operaciones numéricas y trabajar con vectores y matrices. En este ejercicio se utilizará para crear y almacenar la matriz de resultados y para efectuar operaciones matemáticas.

- **Matplotlib (`matplotlib.pyplot`)**: permite construir representaciones gráficas a partir de los resultados obtenidos. En este ejercicio se utilizará para representar la variación del módulo de elasticidad del hormigón, $E_c$, en función de la resistencia especificada a la compresión, $f'_c$.

A continuación se importan las bibliotecas necesarias.

```python
# ============================================================
# IMPORTACIÓN DE BIBLIOTECAS
# ============================================================

# NumPy permite realizar operaciones numéricas y trabajar
# con vectores y matrices.
import numpy as np

# Matplotlib permite construir representaciones gráficas
# de los resultados obtenidos.
import matplotlib.pyplot as plt
```

## 5. Creación de la matriz de resultados

Los valores calculados del módulo de elasticidad del hormigón se almacenarán en una matriz denominada **Ec**.

La matriz tendrá:

- **4 filas**, correspondientes a los cuatro tipos de origen del agregado grueso.
- **101 columnas**, correspondientes a valores de $f'_c$ desde 0 hasta 100 MPa, con incrementos de 1 MPa.

Por lo tanto, la posición **Ec[OA, J]** almacenará el valor del módulo de elasticidad calculado para un origen de agregado determinado, identificado mediante **OA**, y para un valor específico de $f'_c$, identificado mediante **J**.

Inicialmente, todos los valores de la matriz se establecerán iguales a cero.

```python
# ============================================================
# CREACIÓN DE LA MATRIZ DE RESULTADOS
# ============================================================

# Se crea una matriz de 4 filas y 101 columnas.
# Inicialmente, todos sus valores son iguales a cero.
Ec = np.zeros((4, 101))

# Se muestran las dimensiones de la matriz creada.
print("Dimensiones de la matriz Ec:", Ec.shape)
```

## 6. Selección del coeficiente según el origen del agregado

El procedimiento debe realizar los cálculos para los cuatro posibles orígenes del agregado grueso. Para ello se utilizará una estructura repetitiva que permitirá recorrer los valores de **OA** desde 0 hasta 3.

Para cada valor de **OA**, el programa deberá seleccionar el coeficiente **K** correspondiente. Esta selección se realizará mediante una estructura condicional:

- Si **OA = 0**, el agregado es de origen ígneo y **K = 5500**.
- Si **OA = 1**, el agregado es de origen metamórfico y **K = 4700**.
- Si **OA = 2**, el agregado es de origen sedimentario y **K = 3600**.
- En cualquier otro caso considerado en este ejercicio, **K = 3900**, correspondiente al origen del agregado sin distinción.

De esta manera, una decisión ingenieril expresada inicialmente mediante cuatro ecuaciones diferentes se transforma en una estructura lógica que puede ser ejecutada por el computador.

```python
# ============================================================
# SELECCIÓN DEL COEFICIENTE K
# ============================================================

# El ciclo permite recorrer los cuatro tipos de agregado.
for I in range(4):

    # Se asigna el valor del contador I a la variable OA.
    OA = I

    # Se selecciona el coeficiente K según el origen
    # del agregado grueso.
    if OA == 0:
        K = 5500
    elif OA == 1:
        K = 4700
    elif OA == 2:
        K = 3600
    else:
        K = 3900

    # Se muestran los valores seleccionados.
    print("OA =", OA, "   K =", K)
```

## 7. Cálculo del módulo de elasticidad

Una vez seleccionado el coeficiente **K** correspondiente al origen del agregado grueso, se calculará el módulo de elasticidad para valores de $f'_c$ comprendidos entre 0 y 100 MPa, con incrementos de 1 MPa.

Para cada origen del agregado, identificado mediante **OA**, se realizará un segundo ciclo controlado por la variable **J**.

En este ejercicio se hará coincidir el valor de **J** con el valor de $f'_c$:

$$
f'_c = J
$$

Para cada combinación de **OA** y **J**, el módulo de elasticidad se calculará mediante:

$$
E_c = K\sqrt{f'_c}
$$

El resultado obtenido se almacenará en la posición correspondiente de la matriz:

$$
Ec[OA,J]
$$

De esta manera, el ciclo exterior recorre los tipos de agregado y el ciclo interior recorre los valores de resistencia especificada a la compresión.

```python
# ============================================================
# CÁLCULO DEL MÓDULO DE ELASTICIDAD
# ============================================================

# El ciclo exterior recorre los cuatro tipos de agregado.
for I in range(4):

    OA = I

    # Selección del coeficiente K.
    if OA == 0:
        K = 5500
    elif OA == 1:
        K = 4700
    elif OA == 2:
        K = 3600
    else:
        K = 3900

    # El ciclo interior recorre valores de f'c entre
    # 0 y 100 MPa.
    for J in range(101):

        Fpc = J

        # Cálculo y almacenamiento del módulo de elasticidad.
        Ec[OA, J] = K * np.sqrt(Fpc)
```

## 8. Verificación de los resultados almacenados

Aunque la celda anterior no produjo una salida visible, el programa ejecutó los cálculos y almacenó los resultados en la matriz **Ec**.

Para verificar su contenido se consultarán algunos valores específicos de la matriz.

Como ejemplo, se evaluarán los resultados correspondientes a:

$$
f'_c = 15 \text{ MPa}
$$

para los cuatro tipos de agregado considerados.

```python
# ============================================================
# VERIFICACIÓN DE LOS RESULTADOS ALMACENADOS
# ============================================================

Fpc_consulta = 15

print("Resultados para f'c =", Fpc_consulta, "MPa\n")

print("Agregado ígneo:        Ec =", Ec[0, Fpc_consulta], "MPa")
print("Agregado metamórfico:  Ec =", Ec[1, Fpc_consulta], "MPa")
print("Agregado sedimentario: Ec =", Ec[2, Fpc_consulta], "MPa")
print("Sin distinción:        Ec =", Ec[3, Fpc_consulta], "MPa")
```

## 9. Representación gráfica de los resultados

Una representación gráfica permite comparar el efecto del origen del agregado grueso sobre el módulo de elasticidad estimado del hormigón.

En la gráfica se representarán los valores de $f'_c$ comprendidos entre 0 y 100 MPa y, para cada uno de los cuatro tipos de agregado considerados, los valores de $E_c$ previamente calculados y almacenados en la matriz **Ec**.

El eje horizontal representará la resistencia especificada a la compresión del hormigón, $f'_c$, mientras que el eje vertical representará el módulo de elasticidad, $E_c$.

```python
# ============================================================
# REPRESENTACIÓN GRÁFICA DE LOS RESULTADOS
# ============================================================

# Valores de f'c correspondientes a las columnas de la matriz Ec.
Fpc_grafica = np.arange(0, 101, 1)

# Se crea la figura.
plt.figure(figsize=(10, 6))

# Se representan las cuatro filas de la matriz Ec.
plt.plot(Fpc_grafica, Ec[0, :], label="Ígneo")
plt.plot(Fpc_grafica, Ec[1, :], label="Metamórfico")
plt.plot(Fpc_grafica, Ec[2, :], label="Sedimentario")
plt.plot(Fpc_grafica, Ec[3, :], label="Sin distinción")

# Identificación de los ejes y de la gráfica.
plt.xlabel("Resistencia especificada a la compresión, f'c [MPa]")
plt.ylabel("Módulo de elasticidad del hormigón, Ec [MPa]")
plt.title("Módulo de elasticidad del hormigón")

# Límites de representación.
plt.xlim(0, 100)
plt.ylim(bottom=0)

# Divisiones del eje horizontal cada 5 MPa.
plt.xticks(np.arange(0, 101, 5))

# Elementos auxiliares de lectura.
plt.grid(True)
plt.legend()

# Se muestra la gráfica.
plt.show()
```
