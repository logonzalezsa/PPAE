## 1. Planteamiento del problema

En Resistencia de Materiales pueden plantearse dos tipos de problemas complementarios:

### Análisis

Se conocen la **solicitación** y las **propiedades geométricas del elemento**, y se determina la respuesta mecánica correspondiente.

$$
\boxed{
\text{Solicitación}
+
\text{Geometría}
\longrightarrow
\text{Respuesta}
}
$$

### Dimensionamiento

Se conocen la **solicitación** y el **valor admisible de la respuesta**, y se determina la propiedad geométrica mínima requerida.

$$
\boxed{
\text{Solicitación}
+
\text{Respuesta admisible}
\longrightarrow
\text{Propiedad geométrica requerida}
}
$$

En este cuaderno se desarrollará un algoritmo general que permita seleccionar el tipo de solicitación y posteriormente decidir si se desea realizar un problema de **análisis** o de **dimensionamiento**.

## 5. Tracción axial

Para un elemento sometido a una fuerza axial de tracción $F$, el esfuerzo normal promedio se determina mediante:

$$
\sigma=\frac{F}{A}
$$

donde:

- $F$ = fuerza axial aplicada.
- $A$ = área de la sección transversal.
- $\sigma$ = esfuerzo normal.

La misma expresión permite plantear dos problemas diferentes.

### Análisis

Si se conocen $F$ y $A$:

$$
\boxed{\sigma=\frac{F}{A}}
$$

### Dimensionamiento

Si se conocen $F$ y el esfuerzo admisible $\sigma_{adm}$:

$$
\boxed{A_{req}=\frac{F}{\sigma_{adm}}}
$$

El área calculada representa el **área mínima requerida** para satisfacer el criterio de esfuerzo adoptado.

### Sistemas de unidades

En el desarrollo del ejercicio se emplearán sistemas de unidades consistentes:

| Sistema | Fuerza | Área | Esfuerzo |
|---|---:|---:|---:|
| SI | N | mm² | MPa |
| Inglés | lb | in² | psi |

Recordando que:

$$
1\;MPa=1\;\frac{N}{mm^2}
$$

y:

$$
1\;psi=1\;\frac{lb}{in^2}
$$

El algoritmo solicitará inicialmente el sistema de unidades y utilizará las unidades correspondientes en la entrada y presentación de los resultados.

```python
# ============================================================
# TRACCIÓN AXIAL
# ============================================================

if solicitacion == 1:

    print("\nTRACCIÓN AXIAL")
    print("==============")

    # --------------------------------------------------------
    # SELECCIÓN DEL SISTEMA DE UNIDADES
    # --------------------------------------------------------

    print("\nSISTEMA DE UNIDADES")
    print("1. SI      → N, mm², MPa")
    print("2. Inglés  → lb, in², psi")

    while True:
        try:
            unidades = int(
                input("\nSeleccione el sistema de unidades (1-2): ")
            )

            if unidades in [1, 2]:
                break
            else:
                print("Opción no válida. Ingrese 1 o 2.")

        except ValueError:
            print("Entrada no válida.")

    if unidades == 1:
        uF = "N"
        uA = "mm²"
        uSigma = "MPa"

    else:
        uF = "lb"
        uA = "in²"
        uSigma = "psi"

    # --------------------------------------------------------
    # FUERZA APLICADA
    # --------------------------------------------------------

    F = float(
        input(f"\nIngrese la fuerza axial F [{uF}]: ")
    )

    # --------------------------------------------------------
    # ANÁLISIS
    # --------------------------------------------------------

    if problema == 1:

        A = float(
            input(f"Ingrese el área de la sección A [{uA}]: ")
        )

        sigma = F / A

        print("\nRESULTADO DEL ANÁLISIS")
        print("======================")
        print(f"F = {F:.4f} {uF}")
        print(f"A = {A:.4f} {uA}")
        print(f"σ = {sigma:.4f} {uSigma}")

    # --------------------------------------------------------
    # DIMENSIONAMIENTO
    # --------------------------------------------------------

    elif problema == 2:

        sigma_adm = float(
            input(
                f"Ingrese el esfuerzo admisible σadm [{uSigma}]: "
            )
        )

        A_req = F / sigma_adm

        print("\nRESULTADO DEL DIMENSIONAMIENTO")
        print("==============================")
        print(f"F       = {F:.4f} {uF}")
        print(f"σadm    = {sigma_adm:.4f} {uSigma}")
        print(f"A mín.  = {A_req:.4f} {uA}")

else:

    print(
        "La solicitación seleccionada no corresponde "
        "a tracción axial."
    )
```

## 6. Cortante directo

Cuando una fuerza produce cortante directo sobre un elemento, el esfuerzo cortante promedio depende del número de planos resistentes.

La expresión general puede escribirse como:

$$
\tau=\frac{F}{nA}
$$

donde:

- $F$ = fuerza cortante aplicada.
- $A$ = área resistente correspondiente a cada plano de corte.
- $n$ = número de planos resistentes.
- $\tau$ = esfuerzo cortante promedio.

Para **cortante simple**:

$$
n=1
$$

y:

$$
\tau=\frac{F}{A}
$$

Para **cortante doble**:

$$
n=2
$$

y:

$$
\tau=\frac{F}{2A}
$$

### Análisis

Conocidos $F$, $A$ y el número de planos resistentes:

$$
\boxed{\tau=\frac{F}{nA}}
$$

### Dimensionamiento

Conocidos $F$, $\tau_{adm}$ y el número de planos resistentes:

$$
\boxed{A_{req}=\frac{F}{n\tau_{adm}}}
$$

El área obtenida corresponde al **área mínima requerida en cada plano resistente**.

### Sistemas de unidades

| Sistema | Fuerza | Área por plano | Esfuerzo cortante |
|---|---:|---:|---:|
| SI | N | mm² | MPa |
| Inglés | lb | in² | psi |

```python
# ============================================================
# CORTANTE DIRECTO
# ============================================================

if solicitacion == 2:

    print("\nCORTANTE DIRECTO")
    print("================")

    # --------------------------------------------------------
    # SELECCIÓN DEL SISTEMA DE UNIDADES
    # --------------------------------------------------------

    print("\nSISTEMA DE UNIDADES")
    print("1. SI      → N, mm², MPa")
    print("2. Inglés  → lb, in², psi")

    while True:
        try:
            unidades = int(
                input("\nSeleccione el sistema de unidades (1-2): ")
            )

            if unidades in [1, 2]:
                break
            else:
                print("Opción no válida. Ingrese 1 o 2.")

        except ValueError:
            print("Entrada no válida. Debe ingresar un número entero.")

    if unidades == 1:
        uF = "N"
        uA = "mm²"
        uTau = "MPa"

    else:
        uF = "lb"
        uA = "in²"
        uTau = "psi"

    # --------------------------------------------------------
    # SELECCIÓN DEL TIPO DE CORTANTE
    # --------------------------------------------------------

    print("\nTIPO DE CORTANTE")
    print("1. Cortante simple")
    print("2. Cortante doble")

    while True:
        try:
            tipo_cortante = int(
                input("\nSeleccione el tipo de cortante (1-2): ")
            )

            if tipo_cortante in [1, 2]:
                break
            else:
                print("Opción no válida. Ingrese 1 o 2.")

        except ValueError:
            print("Entrada no válida. Debe ingresar un número entero.")

    if tipo_cortante == 1:
        n_planos = 1
        nombre_cortante = "Cortante simple"

    else:
        n_planos = 2
        nombre_cortante = "Cortante doble"

    print(f"\nTipo seleccionado: {nombre_cortante}")
    print(f"Número de planos resistentes: {n_planos}")

    # --------------------------------------------------------
    # FUERZA CORTANTE
    # --------------------------------------------------------

    F = float(
        input(f"\nIngrese la fuerza cortante F [{uF}]: ")
    )

    # --------------------------------------------------------
    # ANÁLISIS
    # --------------------------------------------------------

    if problema == 1:

        A = float(
            input(
                f"Ingrese el área resistente por plano A [{uA}]: "
            )
        )

        tau = F / (n_planos * A)

        print("\nRESULTADO DEL ANÁLISIS")
        print("======================")
        print(f"Tipo        = {nombre_cortante}")
        print(f"N.º planos  = {n_planos}")
        print(f"F           = {F:.4f} {uF}")
        print(f"A por plano = {A:.4f} {uA}")
        print(f"τ           = {tau:.4f} {uTau}")

    # --------------------------------------------------------
    # DIMENSIONAMIENTO
    # --------------------------------------------------------

    elif problema == 2:

        tau_adm = float(
            input(
                f"Ingrese el esfuerzo cortante admisible "
                f"τadm [{uTau}]: "
            )
        )

        A_req = F / (n_planos * tau_adm)

        print("\nRESULTADO DEL DIMENSIONAMIENTO")
        print("==============================")
        print(f"Tipo        = {nombre_cortante}")
        print(f"N.º planos  = {n_planos}")
        print(f"F           = {F:.4f} {uF}")
        print(f"τadm        = {tau_adm:.4f} {uTau}")
        print(f"A mín/plano = {A_req:.4f} {uA}")

else:

    print(
        "La solicitación seleccionada no corresponde "
        "a cortante directo."
    )
```

## 7. Torsión

Para un eje circular sometido a un momento torsor $T$, el esfuerzo cortante máximo se determina mediante:

$$
\tau_{max}=\frac{Tr}{J}
$$

donde:

- $T$ = momento torsor aplicado.
- $r$ = distancia desde el centro hasta la fibra exterior.
- $J$ = momento polar de inercia.
- $\tau_{max}$ = esfuerzo cortante máximo.

En este ejercicio se considerarán dos tipos de sección circular.

### 1. Sección circular maciza

Para un diámetro $d$:

$$
J=\frac{\pi d^4}{32}
$$

y:

$$
\boxed{
\tau_{max}=\frac{16T}{\pi d^3}
}
$$

Para dimensionamiento:

$$
\boxed{
d_{req}=
\sqrt[3]{\frac{16T}{\pi\tau_{adm}}}
}
$$

### 2. Sección circular hueca

Para un diámetro exterior $D$ y un diámetro interior $d$:

$$
J=\frac{\pi}{32}(D^4-d^4)
$$

y:

$$
\boxed{
\tau_{max}=
\frac{16TD}{\pi(D^4-d^4)}
}
$$

Para dimensionamiento se define la relación:

$$
\alpha=\frac{d}{D}
$$

de manera que:

$$
d=\alpha D
$$

y se obtiene:

$$
\boxed{
D_{req}=
\sqrt[3]{
\frac{16T}
{\pi\tau_{adm}(1-\alpha^4)}
}
}
$$

Una vez calculado $D_{req}$:

$$
d_{req}=\alpha D_{req}
$$

### Sistemas de unidades

| Sistema | Momento torsor | Diámetro | Esfuerzo |
|---|---:|---:|---:|
| SI | N·mm | mm | MPa |
| Inglés | lb·in | in | psi |

Las unidades deben mantenerse consistentes durante todo el cálculo.

```python
# ============================================================
# TORSIÓN
# ============================================================

import numpy as np

if solicitacion == 3:

    print("\nTORSIÓN")
    print("=======")

    # --------------------------------------------------------
    # SISTEMA DE UNIDADES
    # --------------------------------------------------------

    print("\nSISTEMA DE UNIDADES")
    print("1. SI      → N·mm, mm, MPa")
    print("2. Inglés  → lb·in, in, psi")

    while True:
        try:
            unidades = int(
                input("\nSeleccione el sistema de unidades (1-2): ")
            )

            if unidades in [1, 2]:
                break
            else:
                print("Opción no válida. Ingrese 1 o 2.")

        except ValueError:
            print("Entrada no válida. Debe ingresar un número entero.")

    if unidades == 1:
        uT = "N·mm"
        uD = "mm"
        uTau = "MPa"

    else:
        uT = "lb·in"
        uD = "in"
        uTau = "psi"

    # --------------------------------------------------------
    # TIPO DE SECCIÓN
    # --------------------------------------------------------

    print("\nTIPO DE SECCIÓN")
    print("1. Circular maciza")
    print("2. Circular hueca")

    while True:
        try:
            tipo_seccion = int(
                input("\nSeleccione el tipo de sección (1-2): ")
            )

            if tipo_seccion in [1, 2]:
                break
            else:
                print("Opción no válida. Ingrese 1 o 2.")

        except ValueError:
            print("Entrada no válida. Debe ingresar un número entero.")

    # --------------------------------------------------------
    # MOMENTO TORSOR
    # --------------------------------------------------------

    T = float(
        input(f"\nIngrese el momento torsor T [{uT}]: ")
    )

    # ========================================================
    # SECCIÓN CIRCULAR MACIZA
    # ========================================================

    if tipo_seccion == 1:

        nombre_seccion = "Circular maciza"

        # ----------------------------------------------------
        # ANÁLISIS
        # ----------------------------------------------------

        if problema == 1:

            d = float(
                input(f"Ingrese el diámetro d [{uD}]: ")
            )

            J = np.pi * d**4 / 32
            tau = 16 * T / (np.pi * d**3)

            print("\nRESULTADO DEL ANÁLISIS")
            print("======================")
            print(f"Sección = {nombre_seccion}")
            print(f"T       = {T:.4f} {uT}")
            print(f"d       = {d:.4f} {uD}")
            print(f"J       = {J:.4f} {uD}⁴")
            print(f"τmax    = {tau:.4f} {uTau}")

        # ----------------------------------------------------
        # DIMENSIONAMIENTO
        # ----------------------------------------------------

        elif problema == 2:

            tau_adm = float(
                input(
                    f"Ingrese el esfuerzo cortante admisible "
                    f"τadm [{uTau}]: "
                )
            )

            d_req = (
                16 * T / (np.pi * tau_adm)
            ) ** (1/3)

            J_req = np.pi * d_req**4 / 32

            print("\nRESULTADO DEL DIMENSIONAMIENTO")
            print("==============================")
            print(f"Sección = {nombre_seccion}")
            print(f"T       = {T:.4f} {uT}")
            print(f"τadm    = {tau_adm:.4f} {uTau}")
            print(f"d mín.  = {d_req:.4f} {uD}")
            print(f"J       = {J_req:.4f} {uD}⁴")

    # ========================================================
    # SECCIÓN CIRCULAR HUECA
    # ========================================================

    elif tipo_seccion == 2:

        nombre_seccion = "Circular hueca"

        # ----------------------------------------------------
        # ANÁLISIS
        # ----------------------------------------------------

        if problema == 1:

            D = float(
                input(f"Ingrese el diámetro exterior D [{uD}]: ")
            )

            while True:

                d = float(
                    input(
                        f"Ingrese el diámetro interior d [{uD}]: "
                    )
                )

                if 0 <= d < D:
                    break
                else:
                    print(
                        "Debe cumplirse: 0 ≤ d < D."
                    )

            J = np.pi * (D**4 - d**4) / 32

            tau = (
                16 * T * D /
                (np.pi * (D**4 - d**4))
            )

            print("\nRESULTADO DEL ANÁLISIS")
            print("======================")
            print(f"Sección = {nombre_seccion}")
            print(f"T       = {T:.4f} {uT}")
            print(f"D       = {D:.4f} {uD}")
            print(f"d       = {d:.4f} {uD}")
            print(f"J       = {J:.4f} {uD}⁴")
            print(f"τmax    = {tau:.4f} {uTau}")

        # ----------------------------------------------------
        # DIMENSIONAMIENTO
        # ----------------------------------------------------

        elif problema == 2:

            tau_adm = float(
                input(
                    f"Ingrese el esfuerzo cortante admisible "
                    f"τadm [{uTau}]: "
                )
            )

            print(
                "\nDefina la relación geométrica α = d/D"
            )

            while True:

                alpha = float(
                    input(
                        "Ingrese α (0 ≤ α < 1): "
                    )
                )

                if 0 <= alpha < 1:
                    break
                else:
                    print(
                        "La relación α debe estar entre 0 y 1."
                    )

            D_req = (
                16 * T /
                (
                    np.pi *
                    tau_adm *
                    (1 - alpha**4)
                )
            ) ** (1/3)

            d_req = alpha * D_req

            J_req = (
                np.pi *
                (D_req**4 - d_req**4) /
                32
            )

            print("\nRESULTADO DEL DIMENSIONAMIENTO")
            print("==============================")
            print(f"Sección = {nombre_seccion}")
            print(f"T       = {T:.4f} {uT}")
            print(f"τadm    = {tau_adm:.4f} {uTau}")
            print(f"α=d/D   = {alpha:.4f}")
            print(f"D mín.  = {D_req:.4f} {uD}")
            print(f"d       = {d_req:.4f} {uD}")
            print(f"J       = {J_req:.4f} {uD}⁴")

else:

    print(
        "La solicitación seleccionada no corresponde a torsión."
    )
```

## 8. Pandeo de columnas

Un elemento esbelto sometido a compresión axial puede perder su estabilidad antes de alcanzar la resistencia del material. Este fenómeno se denomina **pandeo**.

Para una columna ideal, la carga crítica elástica de Euler se expresa mediante:

$$
F_{cr}=
\frac{\pi^2EI}{L_e^2}
$$

donde la longitud efectiva es:

$$
L_e=kL
$$

Por tanto:

$$
\boxed{
F_{cr}=
\frac{\pi^2EI}{(kL)^2}
}
$$

donde:

- $F_{cr}$ = carga crítica de Euler.
- $E$ = módulo de elasticidad del material.
- $I$ = momento de inercia de la sección respecto al eje de pandeo.
- $L$ = longitud real del elemento.
- $k$ = factor de longitud efectiva.
- $L_e$ = longitud efectiva de pandeo.

### Condiciones idealizadas de apoyo

| Condición de apoyo | $k$ |
|---|---:|
| Articulado – articulado | 1.0 |
| Empotrado – libre | 2.0 |
| Empotrado – articulado | 0.7 |
| Empotrado – empotrado | 0.5 |

### Análisis

Conocidos $E$, $I$, $L$ y la condición de apoyo:

$$
\boxed{
F_{cr}=
\frac{\pi^2EI}{(kL)^2}
}
$$

### Dimensionamiento

Si se establece una carga axial $F$ que el elemento debe alcanzar como carga crítica mínima:

$$
F=
\frac{\pi^2EI_{req}}{(kL)^2}
$$

de donde:

$$
\boxed{
I_{req}=
\frac{F(kL)^2}{\pi^2E}
}
$$

El valor obtenido corresponde al **momento de inercia mínimo requerido respecto al eje alrededor del cual puede producirse el pandeo**.

> **Nota:** la ecuación de Euler representa un modelo ideal de inestabilidad elástica. En este ejercicio se emplea con fines académicos para estudiar la relación entre carga, rigidez, longitud y condiciones de apoyo. No se incorporan factores de seguridad ni verificaciones normativas de diseño.

### Sistemas de unidades

| Sistema | Fuerza | Longitud | $E$ | $I$ |
|---|---:|---:|---:|---:|
| SI | N | mm | MPa | mm⁴ |
| Inglés | lb | in | psi | in⁴ |

Las unidades deben mantenerse consistentes durante todo el cálculo.

```python
# ============================================================
# PANDEO DE COLUMNAS - ECUACIÓN DE EULER
# ============================================================

import numpy as np

if solicitacion == 4:

    print("\nPANDEO DE COLUMNAS")
    print("==================")

    # --------------------------------------------------------
    # SISTEMA DE UNIDADES
    # --------------------------------------------------------

    print("\nSISTEMA DE UNIDADES")
    print("1. SI      → N, mm, MPa, mm⁴")
    print("2. Inglés  → lb, in, psi, in⁴")

    while True:
        try:
            unidades = int(
                input("\nSeleccione el sistema de unidades (1-2): ")
            )

            if unidades in [1, 2]:
                break
            else:
                print("Opción no válida. Ingrese 1 o 2.")

        except ValueError:
            print("Entrada no válida. Debe ingresar un número entero.")

    if unidades == 1:
        uF = "N"
        uL = "mm"
        uE = "MPa"
        uI = "mm⁴"

    else:
        uF = "lb"
        uL = "in"
        uE = "psi"
        uI = "in⁴"

    # --------------------------------------------------------
    # CONDICIÓN DE APOYO
    # --------------------------------------------------------

    print("\nCONDICIÓN DE APOYO")
    print("1. Articulado - articulado")
    print("2. Empotrado - libre")
    print("3. Empotrado - articulado")
    print("4. Empotrado - empotrado")

    while True:
        try:
            apoyo = int(
                input("\nSeleccione la condición de apoyo (1-4): ")
            )

            if apoyo in [1, 2, 3, 4]:
                break
            else:
                print("Opción no válida. Ingrese un número entre 1 y 4.")

        except ValueError:
            print("Entrada no válida. Debe ingresar un número entero.")

    condiciones = {
        1: ("Articulado - articulado", 1.0),
        2: ("Empotrado - libre", 2.0),
        3: ("Empotrado - articulado", 0.7),
        4: ("Empotrado - empotrado", 0.5)
    }

    nombre_apoyo, k = condiciones[apoyo]

    print(f"\nCondición seleccionada: {nombre_apoyo}")
    print(f"Factor de longitud efectiva k = {k}")

    # --------------------------------------------------------
    # DATOS COMUNES
    # --------------------------------------------------------

    E = float(
        input(f"\nIngrese el módulo de elasticidad E [{uE}]: ")
    )

    L = float(
        input(f"Ingrese la longitud real L [{uL}]: ")
    )

    Le = k * L

    # ========================================================
    # ANÁLISIS
    # ========================================================

    if problema == 1:

        I = float(
            input(
                f"Ingrese el momento de inercia I [{uI}]: "
            )
        )

        Fcr = (
            np.pi**2 * E * I /
            Le**2
        )

        print("\nRESULTADO DEL ANÁLISIS")
        print("======================")
        print(f"Apoyo   = {nombre_apoyo}")
        print(f"k       = {k:.2f}")
        print(f"L       = {L:.4f} {uL}")
        print(f"Le      = {Le:.4f} {uL}")
        print(f"E       = {E:.4f} {uE}")
        print(f"I       = {I:.4f} {uI}")
        print(f"Fcr     = {Fcr:.4f} {uF}")

    # ========================================================
    # DIMENSIONAMIENTO
    # ========================================================

    elif problema == 2:

        F = float(
            input(
                f"Ingrese la carga crítica mínima requerida F [{uF}]: "
            )
        )

        I_req = (
            F * Le**2 /
            (np.pi**2 * E)
        )

        print("\nRESULTADO DEL DIMENSIONAMIENTO")
        print("==============================")
        print(f"Apoyo   = {nombre_apoyo}")
        print(f"k       = {k:.2f}")
        print(f"L       = {L:.4f} {uL}")
        print(f"Le      = {Le:.4f} {uL}")
        print(f"E       = {E:.4f} {uE}")
        print(f"F mín.  = {F:.4f} {uF}")
        print(f"I mín.  = {I_req:.4f} {uI}")

else:

    print(
        "La solicitación seleccionada no corresponde a pandeo."
    )
```

## 9. Síntesis del algoritmo

El cuaderno desarrollado integra diferentes problemas de Resistencia de Materiales dentro de una misma estructura computacional.

La primera decisión corresponde al **tipo de solicitación**:

$$
\boxed{
\text{Tracción}
\;|\;
\text{Cortante}
\;|\;
\text{Torsión}
\;|\;
\text{Pandeo}
}
$$

La segunda decisión establece el propósito del cálculo:

$$
\boxed{
\text{Análisis}
\quad\rightleftarrows\quad
\text{Dimensionamiento}
}
$$

De esta manera se generan las siguientes rutas principales:

| Solicitación | Análisis | Dimensionamiento |
|---|---|---|
| Tracción | $F,A\rightarrow\sigma$ | $F,\sigma_{adm}\rightarrow A_{req}$ |
| Cortante directo | $F,n,A\rightarrow\tau$ | $F,n,\tau_{adm}\rightarrow A_{req}$ |
| Torsión | $T,\text{geometría}\rightarrow\tau_{max}$ | $T,\tau_{adm}\rightarrow\text{geometría requerida}$ |
| Pandeo | $E,I,L,k\rightarrow F_{cr}$ | $F,E,L,k\rightarrow I_{req}$ |

Algunas rutas requieren decisiones adicionales:

- En **cortante directo**, se identifica si existe cortante simple o doble.
- En **torsión**, se selecciona una sección circular maciza o hueca.
- En **pandeo**, se establece la condición idealizada de apoyo y, a partir de ella, el factor de longitud efectiva $k$.
- En cada problema se selecciona un sistema coherente de unidades.

La estructura general del algoritmo puede resumirse como:

$$
\boxed{
\text{Solicitación}
\rightarrow
\text{Tipo de problema}
\rightarrow
\text{Decisiones particulares}
\rightarrow
\text{Datos de entrada}
\rightarrow
\text{Modelo mecánico}
\rightarrow
\text{Resultado}
}
$$

El programa no utiliza una única secuencia rígida de operaciones. La ruta de cálculo depende de las decisiones adoptadas para representar el problema mecánico.

## 10. Reflexión final

Los problemas desarrollados muestran que una ecuación de Resistencia de Materiales puede utilizarse en dos direcciones.

En un problema de **análisis**, las propiedades del elemento son conocidas y se determina su respuesta:

$$
\boxed{
\text{Solicitación}
+
\text{Geometría}
\longrightarrow
\text{Respuesta mecánica}
}
$$

En un problema de **dimensionamiento**, el razonamiento se invierte: se establece una respuesta límite o una condición requerida y se determina la propiedad geométrica necesaria:

$$
\boxed{
\text{Solicitación}
+
\text{Condición requerida}
\longrightarrow
\text{Geometría necesaria}
}
$$

Por tanto, el cambio entre análisis y dimensionamiento no implica necesariamente utilizar un modelo mecánico diferente, sino **interpretar y reorganizar las mismas relaciones físicas de acuerdo con la incógnita del problema**.

La programación permite sistematizar este proceso mediante estructuras de decisión. El algoritmo identifica la solicitación, selecciona la formulación correspondiente, solicita únicamente los datos necesarios y presenta el resultado utilizando unidades consistentes.

Sin embargo, seleccionar una opción dentro de un programa no sustituye la interpretación ingenieril. Antes de realizar cualquier cálculo es necesario reconocer correctamente el fenómeno físico, las condiciones geométricas, los apoyos y las hipótesis asociadas al modelo utilizado.

En particular, las expresiones empleadas en este cuaderno corresponden a modelos simplificados de tracción axial, cortante directo, torsión de ejes circulares y pandeo elástico ideal de Euler. Su aplicación requiere reconocer las condiciones bajo las cuales dichos modelos representan adecuadamente el comportamiento del elemento.

De esta manera, la programación deja de utilizarse únicamente como una herramienta para ejecutar operaciones y se convierte en un medio para organizar el razonamiento empleado en la solución de problemas de ingeniería.

> **Programar el análisis estructural no consiste en escoger una fórmula: consiste en identificar el fenómeno, establecer el modelo y organizar una secuencia lógica que conduzca desde los datos hasta una respuesta físicamente interpretable.**
