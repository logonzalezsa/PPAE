## 1. Planteamiento del problema

La deformación axial de un elemento prismático sometido a una fuerza axial puede determinarse mediante:

$$
\boxed{
\delta=\frac{FL}{AE}
}
$$

donde:

- $\delta$ = deformación axial.
- $F$ = fuerza axial.
- $L$ = longitud del elemento.
- $A$ = área de la sección transversal.
- $E$ = módulo de elasticidad del material.

La ecuación permite establecer diferentes problemas dependiendo de la variable desconocida.

### Deformación axial

Conocidos $F$, $L$, $A$ y $E$:

$$
\boxed{
\delta=\frac{FL}{AE}
}
$$

### Fuerza axial

Conocidos $\delta$, $L$, $A$ y $E$:

$$
\boxed{
F=\frac{\delta AE}{L}
}
$$

### Área de la sección

Conocidos $F$, $L$, $\delta$ y $E$:

$$
\boxed{
A=\frac{FL}{E\delta}
}
$$

### Módulo de elasticidad

Conocidos $F$, $L$, $A$ y una deformación límite $\delta_{max}$:

$$
\boxed{
E_{req}=\frac{FL}{A\delta_{max}}
}
$$

En los tres primeros casos, el algoritmo funcionará como una herramienta general: las propiedades requeridas serán introducidas directamente por el usuario.

En el último caso, el módulo calculado será interpretado como el **módulo de elasticidad mínimo requerido** para satisfacer el criterio de deformación establecido. Posteriormente podrá compararse con valores representativos de diferentes familias de materiales metálicos como apoyo para una preselección por rigidez.

## 2. Selección múltiple mediante `match-case`

En ejercicios anteriores se utilizaron estructuras condicionales del tipo `if`, `elif` y `else` para controlar diferentes rutas de ejecución.

Python dispone también de la estructura:

```python
match opcion:

    case 1:
        # Primera alternativa

    case 2:
        # Segunda alternativa

    case 3:
        # Tercera alternativa

    case 4:
        # Cuarta alternativa

## 3. ¿Cómo funciona `match-case`?

La estructura `match-case` funciona como un **selector de rutas**.

El programa evalúa el valor de una variable y ejecuta **únicamente el bloque correspondiente al caso seleccionado**.

```text
                         ┌─────────────────┐
                         │     opcion      │
                         └────────┬────────┘
                                  │
                                  ▼
                         ┌─────────────────┐
                         │  match opcion   │
                         └────────┬────────┘
                                  │
              ┌───────────────────┼───────────────────┐
              │                   │                   │
              ▼                   ▼                   ▼
          ┌────────┐          ┌────────┐          ┌────────┐
          │ case 1 │          │ case 2 │          │ case 3 │
          └───┬────┘          └───┬────┘          └───┬────┘
              │                   │                   │
              ▼                   ▼                   ▼
        Ejecuta código       Ejecuta código       Ejecuta código
        debajo del 1        debajo del 2        debajo del 3

                                  │
                                  ▼
                              ┌────────┐
                              │ case 4 │
                              └───┬────┘
                                  │
                                  ▼
                            Ejecuta código
                            debajo del 4

````python
# ============================================================
# DEFORMACIÓN AXIAL - SELECCIÓN DE LA VARIABLE A DETERMINAR
# ============================================================

print("=" * 62)
print("     DEFORMACIÓN AXIAL DE ELEMENTOS PRISMÁTICOS")
print("=" * 62)

# ------------------------------------------------------------
# Selección del sistema de unidades
# ------------------------------------------------------------

print("\nSISTEMA DE UNIDADES")
print("1. Sistema Internacional: N, mm, mm², MPa")
print("2. Sistema Inglés: lb, in, in², psi")

sistema = int(input("\nSeleccione el sistema de unidades (1 o 2): "))

match sistema:

    case 1:
        uF = "N"
        uL = "mm"
        uA = "mm²"
        uE = "MPa"
        uD = "mm"

    case 2:
        uF = "lb"
        uL = "in"
        uA = "in²"
        uE = "psi"
        uD = "in"

    case _:
        print("\nOpción de sistema de unidades no válida.")
        raise SystemExit


# ------------------------------------------------------------
# Selección de la variable desconocida
# ------------------------------------------------------------

print("\n" + "=" * 62)
print("¿QUÉ VARIABLE DESEA DETERMINAR?")
print("=" * 62)

print("1. Deformación axial, δ")
print("2. Fuerza axial, F")
print("3. Área de la sección transversal, A")
print("4. Módulo de elasticidad requerido, E")

opcion = int(input("\nSeleccione una opción (1, 2, 3 o 4): "))


# ------------------------------------------------------------
# Selección de la ruta de cálculo mediante match-case
# ------------------------------------------------------------

match opcion:

    # ========================================================
    # CASO 1: DEFORMACIÓN AXIAL
    # ========================================================

    case 1:

        print("\n--- CÁLCULO DE LA DEFORMACIÓN AXIAL ---")

        F = float(input(f"Fuerza axial F [{uF}]: "))
        L = float(input(f"Longitud del elemento L [{uL}]: "))
        A = float(input(f"Área de la sección A [{uA}]: "))
        E = float(input(f"Módulo de elasticidad E [{uE}]: "))

        delta = F * L / (A * E)

        print("\nRESULTADO")
        print("-" * 35)
        print(f"Deformación axial δ = {delta:.4f} {uD}")


    # ========================================================
    # CASO 2: FUERZA AXIAL
    # ========================================================

    case 2:

        print("\n--- CÁLCULO DE LA FUERZA AXIAL ---")

        delta = float(input(f"Deformación axial δ [{uD}]: "))
        L = float(input(f"Longitud del elemento L [{uL}]: "))
        A = float(input(f"Área de la sección A [{uA}]: "))
        E = float(input(f"Módulo de elasticidad E [{uE}]: "))

        F = delta * A * E / L

        print("\nRESULTADO")
        print("-" * 35)
        print(f"Fuerza axial F = {F:.4f} {uF}")


    # ========================================================
    # CASO 3: ÁREA DE LA SECCIÓN TRANSVERSAL
    # ========================================================

    case 3:

        print("\n--- CÁLCULO DEL ÁREA DE LA SECCIÓN ---")

        F = float(input(f"Fuerza axial F [{uF}]: "))
        L = float(input(f"Longitud del elemento L [{uL}]: "))
        delta = float(input(f"Deformación axial admisible δ [{uD}]: "))
        E = float(input(f"Módulo de elasticidad E [{uE}]: "))

        A = F * L / (E * delta)

        print("\nRESULTADO")
        print("-" * 35)
        print(f"Área requerida A = {A:.4f} {uA}")

        print("\nSi δ corresponde a la deformación máxima admisible,")
        print("el valor obtenido representa el área mínima requerida")
        print("según el criterio de rigidez axial considerado.")


    # ========================================================
    # CASO 4: MÓDULO DE ELASTICIDAD REQUERIDO
    # ========================================================

    case 4:

        print("\n--- CÁLCULO DEL MÓDULO DE ELASTICIDAD REQUERIDO ---")

        F = float(input(f"Fuerza axial F [{uF}]: "))
        L = float(input(f"Longitud del elemento L [{uL}]: "))
        A = float(input(f"Área de la sección A [{uA}]: "))
        delta = float(input(f"Deformación axial máxima admisible δ [{uD}]: "))

        E_req = F * L / (A * delta)

        print("\nRESULTADO")
        print("-" * 35)
        print(f"Módulo de elasticidad requerido E = {E_req:.4f} {uE}")


        # ----------------------------------------------------
        # Alacena de familias de materiales metálicos
        # ----------------------------------------------------

        if sistema == 1:

            materiales = {
                "Aleaciones de aluminio": 69000,
                "Aleaciones de titanio": 110000,
                "Latón": 100000,
                "Cobre": 117000,
                "Acero inoxidable": 193000,
                "Aceros estructurales": 200000
            }

        else:

            # Valores aproximados equivalentes en psi
            materiales = {
                "Aleaciones de aluminio": 10.0e6,
                "Aleaciones de titanio": 16.0e6,
                "Latón": 14.5e6,
                "Cobre": 17.0e6,
                "Acero inoxidable": 28.0e6,
                "Aceros estructurales": 29.0e6
            }


        print("\n" + "=" * 62)
        print("ALACENA DE FAMILIAS DE MATERIALES METÁLICOS")
        print("=" * 62)

        for material, E_material in materiales.items():

            if E_material >= E_req:
                estado = "CUMPLE"
            else:
                estado = "NO CUMPLE"

            print(
                f"{material:<28} "
                f"E ≈ {E_material:>12,.0f} {uE}   "
                f"{estado}"
            )


        print("\nNOTA:")
        print("La comparación corresponde únicamente a un criterio")
        print("preliminar de rigidez axial. La selección definitiva")
        print("del material requiere considerar sus propiedades")
        print("específicas y los demás criterios de diseño.")


    # ========================================================
    # OPCIÓN NO VÁLIDA
    # ========================================================

    case _:

        print("\nLa opción seleccionada no es válida.")
````

## 4. Interpretación del algoritmo

La ecuación de deformación axial

$$
\delta=\frac{FL}{AE}
$$

puede utilizarse para determinar diferentes variables dependiendo de la información disponible en el problema.

En este ejercicio, la estructura `match-case` permitió organizar cuatro rutas independientes de cálculo:

$$
\boxed{
\begin{array}{ccc}
\text{Opción seleccionada} & \longrightarrow & \text{Variable calculada}\\
1 & \longrightarrow & \delta\\
2 & \longrightarrow & F\\
3 & \longrightarrow & A\\
4 & \longrightarrow & E
\end{array}}
$$

Una vez seleccionada una opción, Python ejecuta únicamente las instrucciones correspondientes a ese `case`.

En los tres primeros casos, el programa funciona como una herramienta general para resolver problemas de deformación axial a partir de las propiedades suministradas por el usuario.

En el cuarto caso, el módulo de elasticidad calculado se utiliza adicionalmente como criterio preliminar de rigidez para comparar diferentes familias de materiales metálicos.

Esta comparación no selecciona un material específico. Identifica familias de materiales cuyo módulo de elasticidad puede satisfacer el requerimiento establecido. La selección posterior de una aleación o acero particular requiere considerar otras propiedades mecánicas y criterios de diseño.
